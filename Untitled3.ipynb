```python
# CRIANDO A LISTA DE PRODUTOS E PREÇOS DA LOJA -

produtos = ["Camiseta Básica", "Calça Jeans" , "Jaqueta Corta-Vento"]
precos = [49.98, 129.90, 219.90]

for i,j in zip(produtos, precos):
  print (f"O produto {i} custa R${j:.2f}")

#CADASTRO DE NOVO PRODUTO

novo_produto = input("Informe o nome do produto")
novo_preco = float(input("Informe o preço do produto"))

produtos.append(novo_produto)
precos.append(novo_preco)

print(f"O produto {novo_produto} foi cadastrado com sucesso!")


#REMOVENDO NOVO PRODUTO

for i,j in zip(produtos, precos):
  print (f"O produto {i} custa R${j:.2f}")

resposta2 = input("Deseja remover novo produto? ")
if resposta2.lower() == "sim":
  produto_removido = produtos.pop()
  preco_removido = precos.pop()
  print(f"Produto removido: {produto_removido} (Preço: {preco_removido})")

if len(produtos) > 0:
    produto_removido = produtos.pop()
    preco_removido = precos.pop()
    print(f"Produto removido: {produto_removido} (Preço: {preco_removido})")
else:
    print("Estoque vazio!")


item = input("Digite o nome do produto que deseja remover: ")

if item in produtos:
    i = produtos.index(item)

    produto_removido = produtos.pop(i)
    preco_removido = precos.pop(i)

else:
    print("Produto não existe no estoque!")








```
